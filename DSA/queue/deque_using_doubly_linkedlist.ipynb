### Define a class Node with instance object member variables prev, item and next.

```python
class Node:
    def __init__(self, prev=None, item=None, next=None):
        self.prev = prev
        self.item = item
        self.next = next
```

### Define a class Deque to implement deque data structure using double linkedlist concept. Define `__init__()` method to initialize front and rear reference variable and item_count variable to keep track of number of elements in the deque.

```python
class Deque:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        self.item_count = 0
```

### Define a method is_empty() to check if the deque is empty in Deque class.

```python
class Deque:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        self.item_count = 0

    def is_empty(self):
        return self.front == None
```

### In Deque class, define insert_front() method to add data at the front of the deque.

```python
class Deque:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        self.item_count = 0

    def is_emtpy(self):
        return self.front == None
    
    def insert_front(self, data):
        new_node = Node(item=data)
        if self.is_emtpy():
            self.rear = new_node
        else:
            new_node.next = self.front
            self.front.prev = new_node
            
        self.front = new_node
        self.item_count += 1
```

### In deque class, define insert_rear() method to add data at the rear of the deque.

```python
class Deque:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = front
        self.item_count = 0

    def is_empty(self):
        return self.front == None
    
    def insert_rear(self, data):
        new_node = Node(item=data)
        if self.is_empty():
            self.front = new_node
        else:
            self.rear.next = new_node
            new_node.prev = self.rear
        self.rear = new_node
        self.item_count += 1
```

### In Deque class, define delete_front() method to remove front element from the deque.

```python
class Deque:
    def __init__(self):
        self.front = None
        self.rear = None
        self.item_count = 1

    def is_emtpy(self):
        return self.front == None
    
    def delete_front(self):
        if not self.is_emtpy():
            if self.front == self.rear:
                self.front = None
                self.rear = None
            else:
                self.front = self.front.next
                self.front.prev = None
            self.item_count -= 1
        else:
            raise IndexError("Deque is empty!")
```

### In deque class, define get_front() method to return front element of the deque.
