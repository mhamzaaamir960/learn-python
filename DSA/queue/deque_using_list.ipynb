### Define a class Deque to implement deque data structure using list. Define `__init__()` method to create an empty list object member of Deque.

```python
class Deque:
    def __init__(self):
        self.items = []
```

### Define a method is_empty() to check if the deque is empty in Deque class.

```python
class Deque:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0
```

###  In Deque class, define insert_front() method to add data at the front end of the deque.

```python
class Deque:
    def __init__(self):
        self.items = []
    
    def insert_front(self, data):
        self.items.insert(0, data)
```

### In Deque class, define insert_rear() method to add data at the rear end of the deque.

```python
class Deque:
    def __init__(self):
        self.items = []

    def insert_rear(self, data):
        self.items.append(data)
```

### In Deque class, define delete_front() method to remove front element from the deque.

```python
class Deque:
    def __init__(self):
        self.items = []
    
    def is_empty(self):
        return len(self.items) == 0

    def delete_front(self):
        if not self.is_empty():
            self.items.pop(0)
        else:
            raise IndexError("Deque is empty!")
```

### In Deque class, define delete_rear() method to remove rear element from the deque.

```python
class Deque:
    def __init__(self):
        self.items = []
    
    def is_emtpy(self):
        return len(self.items) == 0
    
    def delete_rear(self):
        if not self.is_emtpy():
            self.items.pop()
        else:
            raise IndexError("Deque is empty!")
```

### In Deque class, define get_front() method to return front element of the deque.

```python
class Deque:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0
    
    def get_front(self):
        if not self.is_empty():
            return self.items[0]
        else:
            raise IndexError("Deque is empty!")
```

### In Deque class, define get_rear() method to return rear element of the deque.

```python
class Deque:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0
    
    def get_rear(self):
        if not self.is_empty():
            return self.items[-1]
        else:
            raise IndexError("Deque is empty!")
```

### In Deque class, define size() method to return size of the deque that is number of items present in the deque.

```python
class Deque:
    def __init__(self):
        self.items = []
    
    def size(self):
        return len(self.items)
```
