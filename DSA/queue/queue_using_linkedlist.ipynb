### Define a class Queue to implement queue data structure using singly linked list concept. Define `__init__()` method to initialize front and rear reference variable and item_count variable to keep track of number of elements in the queue.

```python
class Queue:
    def __init__(self,front=None, rear=None):
        self.front = front
        self.rear = rear
```

### Define Queue class, define enqueue() method to add data into the queue.

```python
class Node:
    def __init__(self, item=None, next=None):
        self.item = item
        self.next = next

class Queue:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear  = rear
        
    def is_emtpy(self):
        return self.front and self.rear
    
    def enqueue(self, data):
        new_node = Node(data)
        if self.is_emtpy():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = self.rear.next
```

### In Queue class, define dequeue() method to remove front element from the queue.

```python
class Queue:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        
    def is_empty(self):
        return self.front and self.rear
    
    def dequeue(self):
        if not self.is_empty():
            if self.front == self.rear:
                self.front = None
                self.rear = None
            else:
                self.front = self.front.next
        else:
            raise IndexError("Queue Underflow!")
```

### In Queue class, define get_front() method to return front element of the Queue.

```python
class Queue:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        
    def is_empty(self):
        return self.front and self.rear
    
    def get_front(self):
        if not self.is_empty():
            return self.front.item
        else:
            raise IndexError("Queue Underflow!")
```

### In Queue class, define get_rear() method to return rear element of the Queue.

```python
class Queue:
    def __init__(self, front=None, rear=None):
        self.front = front 
        self.rear = rear
        
    def is_empty(self):
        return self.front and self.rear
    
    def get_rear(self):
        if not self.is_empty():
            return self.rear.item
        else:
            raise IndexError("Queue Underflow!")
```

### In Queue class, define size() method to return size of the queue that is number of items present in the Queue.

```python
class Queue:
    def __init__(self, front=None, rear=None):
        self.front = front
        self.rear = rear
        self.item_count = 0
        
    def is_empty(self):
        return self.front and self.rear
    
    def size(self):
        return self.item_count
```
