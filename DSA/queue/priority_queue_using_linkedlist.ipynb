### Define a node class with instance member variables item, priority, and next.

```python
class Node:
    def __init__(self, item=None, priority=None, next=None):
        self.item = item
        self.priority = priority
        self.next = next
```

### Define a class PriorityQueue to implement queue data structure using singly linked list. Provide `__init__()` method to create a start reference variable (initiallay containing None) and item_count variable (initially 0).

```python
class PriorityQueue:
    def __init__(self, start):
        self.start = start
        self.item_count = 0
```

### Define is_empty() method in PriorityQueue class to check if the priority queue is empty.

```python
class PriorityQueue:
    def __init__(self, start=None):
        self.start = start
        self.item_count = 0
        
    def is_empty(self):
        return self.start == None
```

### Define a push method in PriorityQueue class, to insert new data with given priority.

```python
class PriorityQueue:
    def __init__(self, start=None):
        self.start = start 
        self.item_count = 0
    
    def is_empty(self):
        return self.start == None
    
    def push(self, data, priority):
        new_node = Node(data, priority)
        if self.is_empty():
            self.start = new_node
        else:
            temp = self.start
            if temp.priority >= priority:
                new_node.next = temp
                self.start = new_node
            else:
                while temp.next and temp.next.priority < priority:
                    temp = temp.next
                new_node.next = temp.next
                temp.next = new_node
        self.item_count += 1
```

### Define a pop method in PriorityQueue class, which returns the highest priority data stored in the priority queue data structure Raise exception if priority queue is empty.

```python
class PriorityQueue:
    def __init__(self, start=None) -> None:
        self.start = start 
        self.item_count = 0

    def is_empty(self):
        return self.start == None
    
    def pop(self):
        if not self.is_empty():
            data = self.start.item
            self.start = self.start.next
            self.item_count -= 1
            return data
        else:
            raise Exception("Priority Queue is empty!")
        
```

### In class PriorityQueue, define a method size to return the number of elements present in the priority queue.

```python
class PriorityQueue:
    def __init__(self, start=None):
        self.start = start
        self.item_count = 0

    def size(self):
        return self.item_count
```
