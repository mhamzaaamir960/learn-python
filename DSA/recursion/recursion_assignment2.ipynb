### Write a recursive function to calculate sum of first N natural numbers.

```python
def sumN(n):
    if n==0:
        return 0
    return n + sumN(n-1)
print(sumN(10))
```

### Write a recursive function to calculate factorial of a number.

```python
def factorialN(n):
    if n==1:
        return 1
    return n*factorialN(n-1)

print(factorialN(5))
```
