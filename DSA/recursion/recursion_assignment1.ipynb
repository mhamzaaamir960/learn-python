### Write a recursive function to print first N natural numbers.

```python
def printN(n):
    if n>0:
        printN(n-1)
        print(n, end=", ")
        
printN(10)
```

### Write a recursive function to print first N natural numbers in reverse order.

```python
def printN(n):
    if n>0:
        print(n)
        printN(n-1)
        
printN(10)
        
```

### Write a recursive funtion to print first N odd natural numbers.

```python
def printOdd(n):
    if n>0:
        printOdd(n-1)
        print(2*n-1, end=', ')
            
printOdd(10)
```

### Write a recursive function to print first N even natural numbers.

```python
def printEven(n):
    if n>0:
        printEven(n-1)
        print(2*n, end=", ")
        
printEven(10)
```

### Write recursive function to print first N odd natural numbers in reverse order.

```python
def printOddReverse(n):
    if n>0:
        print(2*n-1, end=", ")
        printOddReverse(n-1)
        
printOddReverse(10)
```

### Write a recursive function to print first N even natural numbers in reverse order.

```python
def printEvenReverse(n):
    if n>0:
        print(2*n, end=", ")
        printEvenReverse(n-1)
        
printEvenReverse(10)
```
