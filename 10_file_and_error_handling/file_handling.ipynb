```python
with open("data.csv", "r") as file:
    next(file)
    with open("top_student.csv", "w") as new_file:
        new_file.write("ID, Name, Age, Marks, City \n")
        for line in file:
            id,name,age,marks,city = line.strip().split(",")

            if int(marks) > 85:
                print(marks)
                new_file.write(f"{id}, {name}, {age}, {marks}, {city} \n")
```

```python
with open("top_student.csv", "r") as file:
    next(file)
    total_marks = []
    for line in file:
        id, name, age, marks, city = line.strip().split(",")
        total_marks.append(int(marks))

    print(total_marks)
    print("Average marks: ", sum(total_marks)/len(total_marks))
    
```
